## Project Plan

#### This investigation aims to evaluate how effectively machine learning models can classify stellar objects using observational survey data, contributing insight into automated classification methods used in modern large-scale astronomical surveys and future telescope systems.

Develop and evaluate machine learning models capable of classifying stellar objects using observational astronomical measurements from the Sloan Digital Sky Survey (SDSS DR18).

    Input = Astronomical observational features of stellar objects, including photometric and survey-derived measurements.

    Output = Predicted stellar object classification:
        STAR
        GALAXY
        QSO (Quasar)


### Phase 1 — Dataset Investigation & Preprocessing

Investigate and preprocess the SDSS DR18 dataset prior to model training.

This phase includes:
    Missing value analysis
    Class distribution analysis
    Feature correlation analysis
    Feature distribution analysis
    Feature selection and preprocessing

The purpose of this phase is to better understand feature relationships, dataset structure, and potential preprocessing requirements before controlled model experimentation.


### Phase 2 — Model Training & Experimental Investigation

Train and evaluate multiple supervised machine learning models using controlled preprocessing and evaluation conditions.

The following models will be implemented and compared:
    Logistic Regression
    KNN
    Decision Tree
    Random Forest
    SVM

Performance evaluation will be based on:
    Accuracy
    Precision
    Recall
    F1-score
    Confusion Matrix Analysis

Additional experimental investigations include:
    Feature group comparison
    Scaling experiments
    Hyperparameter experiments
    Misclassification analysis

The highest-performing model will be selected for further analysis.


### Phase 3 — Comparison Against Existing Work 

Compare the performance of the best-performing model against results reported in existing peer-reviewed studies and related SDSS stellar classification literature.

This comparison will evaluate:
    Classification accuracy
    Preprocessing approaches
    Feature selection methods
    Model behavior and limitations
    Overall effectiveness relative to comparable studies

## Data Used

Data for model training will initially use the SDSS DR18 dataset.

DR18 was selected because it provides modern, large-scale astronomical survey data while remaining highly compatible with existing SDSS-IV stellar classification literature and benchmarking methodologies. This enables meaningful comparison against prior machine learning studies using similar photometric feature sets and classification tasks.

Although DR18 is released under SDSS-V, the underlying photometric classification framework remains closely aligned with earlier SDSS-IV data products commonly used in machine learning research.

A dedicated SDSS-V-native dataset was not selected due to its stronger emphasis on:
    Time domain astronomy
    Milky Way spectroscopy
    Black hole mapping
Additionally, SDSS-V-focused datasets currently have:
    Less established ML benchmarking literature
    Fewer comparable classification studies
    More complex data structures and observational pipelines
As a result, SDSS DR18 provides the best balance between modern survey data, accessibility, and comparability with existing peer-reviewed machine learning work.

## Investigation Plan

### DATASET INVESTIGATIONS

    1. Missing Values
    Check for empty, null, or invalid entries in the dataset.

    2. Class Distribution
    Analyze how many objects belong to each class: STAR, GALAXY, and QSO.
    Purpose: determine whether the dataset is balanced and explain possible confusion matrix patterns.

    3. Correlation Analysis
    Identify which features are strongly correlated with each other.
    Purpose: detect redundant variables and understand relationships between measurements.

    4. Feature Distributions
    Analyze how each feature is distributed across the object classes.
    Purpose: determine which measurements help separate STAR, GALAXY, and QSO objects.


### EXPERIMENTAL INVESTIGATIONS

    1. Feature Group Comparison
    Compare model performance using different sets of variables, such as magnitudes, fluxes, Petrosian radii, and all features.

    2. Misclassification Analysis
    Analyze incorrectly classified objects.
    Purpose: determine which classes are confused most often and why.

    3. Scaling Experiment
    Train models with and without feature scaling.
    Purpose: determine which models are sensitive to differences in feature magnitude.

    4. Hyperparameter Experiment
    Change model settings to test how performance changes.
    Purpose: find stronger configurations and understand model behavior.


### MODEL INVESTIGATIONS

    1. Baseline Model Comparison
Train all models using the same preprocessing, train/test split, and evaluation metrics.
Purpose: fairly compare Logistic Regression, KNN, Decision Tree, Random Forest, and SVM.
