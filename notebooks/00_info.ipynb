________Project Plan________

Develop and evaluate machine learning models capable of classifying stellar objects using observational astronomical measurements.

Input = Astronomical observational features of stellar objects, including photometric and survey-derived measurements.

Output = Predicted stellar object classification:
            STAR
            GALAXY
            QSO (Quasar)

_______Phase 1 — Model Training and Evaluation_______

Train multiple supervised machine learning models on the SDSS DR18 dataset and evaluate their classification performance.
The following models will be implemented and compared:
    Logistic Regression
    KNN
    Decision Tree
    Random Forest
    SVM
Performance evaluation will be based on metrics such as:
    Accuracy
    Precision
    Recall
    F1-score
    Confusion Matrix analysis
The highest-performing model will be selected for further analysis.

______Phase 2 — Comparison Against Existing Work______

Compare the performance of the best-performing model against results reported in existing peer-reviewed studies and related SDSS stellar classification literature.

This comparison will evaluate:
    Classification accuracy
    Preprocessing approaches
    Feature selection methods
    Overall model effectiveness relative to comparable studies



Data Used

Data for model training will initially use the SDSS DR18 dataset.

DR18 was selected because it provides modern, large-scale astronomical survey data while remaining highly compatible with existing SDSS-IV stellar classification literature and benchmarking methodologies. This enables meaningful comparison against prior machine learning studies using similar photometric feature sets and classification tasks.

Although DR18 is released under SDSS-V, the underlying photometric classification framework remains closely aligned with earlier SDSS-IV data products commonly used in machine learning research.

A dedicated SDSS-V-native dataset was not selected due to its stronger emphasis on:
    Time domain astronomy
    Milky Way spectroscopy
    Black hole mapping
Additionally, SDSS-V-focused datasets currently have:
    Less established ML benchmarking literature
    Fewer comparable classification studies
    More complex data structures and observational pipelines
As a result, SDSS DR18 provides the best balance between modern survey data, accessibility, and comparability with existing peer-reviewed machine learning work.
